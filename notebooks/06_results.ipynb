# 06 · Results tables and significance

Builds every results table from the finished runs (`configs/results.yaml` says which runs) and
compares every pair of models with a paired bootstrap: 10,000 resamples of each test set, the same
resamples for every model, on the difference in mean F1 over seeds; percentile 95% CI, two-sided p,
Holm over the 36 pairs of a task. Each main-table score is recomputed from the run's saved
predictions before it is used.

| | |
|---|---|
| Hardware | CPU |
| Time | ~3 min |
| Needs | the runs from notebooks 03–05 (their `metrics.json` and `test_predictions.npz`) |
| Writes | `outputs/results/paper/`: `main.{md,tex,csv}`, `bootstrap.{md,csv}`, `zeroshot.md`, `sanity_*.{md,tex}`, `runs.csv` |

A run that is missing is listed and left out of its cell, so the tables also work part-way.
